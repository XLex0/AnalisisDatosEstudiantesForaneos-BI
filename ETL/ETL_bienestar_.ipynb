### Limpieza CSV

```python
np.random.seed(42)

df = pd.read_csv("data/df_general_errores.csv")

# Eliminar filas totalmente vacías
df = df.dropna(how="all")

# Limpiar nombre y apellido
for col in ["nombre", "apellido"]:
    if col in df.columns:
        df[col] = (
            df[col]
            .astype(str)
            .replace({"nan": np.nan, "NaN": np.nan})
            .str.strip()
        )

# Mantener solo registros con nombre y apellido válidos
if {"nombre", "apellido"}.issubset(df.columns):
    mask_valid = df["nombre"].notna() & df["apellido"].notna()
    mask_valid &= df["nombre"] != ""
    mask_valid &= df["apellido"] != ""
    df = df[mask_valid].copy()

# Quitar duplicados globales
df = df.drop_duplicates().reset_index(drop=True)

# Definir columnas numéricas (solo si existen)
numeric_cols = ["longitud", "latitud", "distancia", "semestre", "anio", "edad"]
numeric_cols = [c for c in numeric_cols if c in df.columns]

# El resto se consideran texto
text_cols = [c for c in df.columns if c not in numeric_cols]

# --------- Tratamiento de numéricas: outliers + imputación media ---------

for col in numeric_cols:
    col_num = pd.to_numeric(df[col], errors="coerce")

    not_na = col_num.dropna()
    if len(not_na) >= 5:
        q1 = not_na.quantile(0.25)
        q3 = not_na.quantile(0.75)
        iqr = q3 - q1

        lower = q1 - 1.5 * iqr
        upper = q3 + 1.5 * iqr

        outliers = (col_num < lower) | (col_num > upper)
        col_num[outliers] = np.nan

    mean_val = col_num.mean()
    col_num = col_num.fillna(mean_val)

    df[col] = col_num

# Redondeo para variables que deben ser enteras
for col in ["semestre", "anio", "edad"]:
    if col in df.columns:
        df[col] = df[col].round().astype(int)

# --------- Tratamiento de texto: reemplazo de valores corruptos ---------

invalid_tokens = {
    "###", "!ERROR!", "###CORRUPT###", "@@@", "<NULL>",
    "INVALID", "BROKEN", "{BAD}", "<script>",
    "ERR", "???", "BAD", "∞∞∞", "±§¶", "ERR!!", "$$$",
    "<NULL>", "C0RRUPT3D", "[DAMAGED]",
    "~~~~", "XXX", "BAD"
}

def is_invalid_text(v):
    if pd.isna(v):
        return True
    s = str(v).strip()
    if s == "":
        return True
    if s in invalid_tokens:
        return True
    if len(s) > 80:
        return True
    if set(s) <= set("#@!<>={}[]±§¶∞") and len(s) <= 10:
        return True
    return False

for col in text_cols:
    serie = df[col].astype(object)

    valid_mask = ~serie.map(is_invalid_text)
    invalid_mask = ~valid_mask

    # Si todo es inválido, no tocamos la columna
    if valid_mask.sum() == 0:
        df[col] = serie
        continue

    valid_vals = serie[valid_mask].astype(str)
    dist = valid_vals.value_counts(normalize=True)

    valores = dist.index.to_numpy()
    probs = dist.values

    n_invalid = invalid_mask.sum()
    if n_invalid > 0:
        relleno = np.random.choice(valores, size=n_invalid, p=probs)
        serie.loc[invalid_mask] = relleno

    df[col] = serie

# Quitar duplicados por nombre + apellido
if {"nombre", "apellido"}.issubset(df.columns):
    df = df.drop_duplicates(subset=["nombre", "apellido"]).reset_index(drop=True)

df_general = df.reset_index(drop=True)

```

```python
# Generar identificador único
def generar_hash(nombre, apellido):
    base = f"{nombre}_{apellido}"
    return hashlib.md5(base.encode("utf-8")).hexdigest()

df_general["identifier"] = df_general.apply(
    lambda row: generar_hash(row["nombre"], row["apellido"]),
    axis=1
)
```

### Limpieza xlsx

```python

df = pd.read_excel("data/df_bienestar_errores.xlsx")
print(df.columns)

np.random.seed(42)

# Eliminar filas totalmente vacías
df = df.dropna(how="all")

# Limpiar nombre y apellido
for col in ["nombre", "apellido"]:
    if col in df.columns:
        df[col] = (
            df[col]
            .astype(str)
            .replace({"nan": np.nan, "NaN": np.nan})
            .str.strip()
        )

# Mantener solo registros con nombre y apellido válidos
if {"nombre", "apellido"}.issubset(df.columns):
    mask_valid = df["nombre"].notna() & df["apellido"].notna()
    mask_valid &= df["nombre"] != ""
    mask_valid &= df["apellido"] != ""
    df = df[mask_valid].copy()

# Quitar duplicados por nombre + apellido
df = df.drop_duplicates(subset=["nombre", "apellido"]).reset_index(drop=True)

# Columnas numéricas (solo si existen)
numeric_cols = [
    "distancia",
    "promedio_académico",
    "viajes_origen",
    "indice_bienestar",
    "distancia_universidad",
    "horas_sueño",
    "tiempo_traslado_min",
]
numeric_cols = [c for c in numeric_cols if c in df.columns]

# El resto se consideran texto
text_cols = [c for c in df.columns if c not in numeric_cols]

# --- Tratamiento numérico: outliers + media ---

for col in numeric_cols:
    col_num = pd.to_numeric(df[col], errors="coerce")

    not_na = col_num.dropna()
    if len(not_na) >= 5:
        q1 = not_na.quantile(0.25)
        q3 = not_na.quantile(0.75)
        iqr = q3 - q1

        lower = q1 - 1.5 * iqr
        upper = q3 + 1.5 * iqr

        outliers = (col_num < lower) | (col_num > upper)
        col_num[outliers] = np.nan

    mean_val = col_num.mean()
    col_num = col_num.fillna(mean_val)

    df[col] = col_num

# Redondear viajes_origen a entero
if "viajes_origen" in df.columns:
    df["viajes_origen"] = df["viajes_origen"].round().astype(int)

# --- Reutilizamos is_invalid_text del código anterior ---

def is_invalid_text(v):
    if pd.isna(v):
        return True
    s = str(v).strip()
    if s == "":
        return True
    invalid_tokens = {
        "###", "!ERROR!", "###CORRUPT###", "@@@", "<NULL>",
        "INVALID", "BROKEN", "{BAD}", "<script>",
        "ERR", "???", "BAD", "∞∞∞", "±§¶", "ERR!!", "$$$",
        "<NULL>", "C0RRUPT3D", "[DAMAGED]",
        "~~~~", "XXX", "BAD"
    }
    if s in invalid_tokens:
        return True
    if len(s) > 80:
        return True
    if set(s) <= set("#@!<>={}[]±§¶∞") and len(s) <= 10:
        return True
    return False

# --- Tratamiento de texto: reemplazo de valores corruptos ---

for col in text_cols:
    serie = df[col].astype(object)

    valid_mask = ~serie.map(is_invalid_text)
    invalid_mask = ~valid_mask

    if valid_mask.sum() == 0:
        df[col] = serie
        continue

    valid_vals = serie[valid_mask].astype(str)
    dist = valid_vals.value_counts(normalize=True)

    valores = dist.index.to_numpy()
    probs = dist.values

    n_invalid = invalid_mask.sum()
    if n_invalid > 0:
        relleno = np.random.choice(valores, size=n_invalid, p=probs)
        serie.loc[invalid_mask] = relleno

    df[col] = serie

# Quitar duplicados por nombre + apellido otra vez por si se generaron
df = df.drop_duplicates(subset=["nombre", "apellido"]).reset_index(drop=True)

# (Opcional) generar el mismo identifier para poder hacer joins
df["identifier"] = df.apply(
    lambda row: generar_hash(row["nombre"], row["apellido"]),
    axis=1
)

df_bienestar = df.reset_index(drop=True)
```

```python

from sqlalchemy import create_engine
# MySQL
mysql_engine = create_engine(
    f"mysql+pymysql://{MY_USER}:{MY_PASSWORD}@{MY_HOST}:{MY_PORT}/{MY_DB}"
)

# Postgres
pg_engine = create_engine(
    f"postgresql+psycopg2://{PG_USER}:{PG_PASSWORD}@{PG_HOST}:{PG_PORT}/{PG_DB}"
)


# Estudiante desde MySQL
df_mysql_estudiante = pd.read_sql("SELECT * FROM estudiante", mysql_engine)
df_mysql_estudiante = df_mysql_estudiante.drop_duplicates(subset=["nombre", "apellido"])

```

```python


# ---- dim_universidad ----
dim_universidad = (
    df_general[["universidad", "carrera"]]
    .drop_duplicates()
    .sort_values(["universidad", "carrera"])
    .reset_index(drop=True)
)
dim_universidad["id_universidad"] = np.arange(1, len(dim_universidad) + 1)

# ---- dim_periodo ----
dim_periodo = (
    df_general[["semestre", "anio", "periodo"]]
    .drop_duplicates()
    .sort_values(["semestre", "anio", "periodo"])
    .reset_index(drop=True)
)
dim_periodo["id_periodo"] = np.arange(1, len(dim_periodo) + 1)

# ---- dim_residencia ----

dim_residencia_a = (
    df_general[["zona", "sector", "tipo_vivienda", "codigo_postal", "nombre", "apellido"]]
    .drop_duplicates()
)

dim_residencia_b = (
    df_bienestar[["tiempo_traslado_min", "nombre", "apellido"]]
    .drop_duplicates()
)

dim_residencia = (
    dim_residencia_a.merge(
        dim_residencia_b,
        on=["nombre", "apellido"],
        how="inner"
    )
    .drop(columns=["nombre", "apellido"])
    .drop_duplicates()
)

dim_residencia = dim_residencia.sort_values(
    ["zona", "sector", "tipo_vivienda", "codigo_postal", "tiempo_traslado_min"]
).reset_index(drop=True)
dim_residencia["id_residencia"] = np.arange(1, len(dim_residencia) + 1)

# ---- dim_estudiante ----
# Usamos el identifier de df_general y cruzamos con MySQL (becado)

dim_estudiante = (
    df_general.merge(
        df_mysql_estudiante[["nombre", "apellido", "es_becado"]],
        on=["nombre", "apellido"],
        how="inner"
    )
)

dim_estudiante = (
    dim_estudiante[["identifier", "edad", "genero", "modalidad", "es_becado"]]
    .drop_duplicates()
    .rename(columns={
        "identifier": "id_estudiante",
        "es_becado": "becado"
    })
    .reset_index(drop=True)
)

# ---- dim_origen ----
dim_origen = (
    df_general[["ciudad", "provincia", "longitud", "latitud"]]
    .drop_duplicates()
    .sort_values(["ciudad", "provincia", "longitud", "latitud"])
    .reset_index(drop=True)
)
dim_origen["id_origen"] = np.arange(1, len(dim_origen) + 1)
```

```python


# Ingresos totales por estudiante (MySQL)
query_ingresos_totales = """
    SELECT 
        e.nombre,
        e.apellido,
        SUM(i.ingreso_cantidad) AS ingreso_total
    FROM ingreso i
    JOIN estudiante e ON i.id_estudiante = e.id_estudiante
    GROUP BY e.nombre, e.apellido
"""
df_ingresos_totales = pd.read_sql(query_ingresos_totales, mysql_engine)

# Egresos totales por estudiante (Postgres)
query_egresos_totales = """
    SELECT 
        e.nombre,
        e.apellido,
        SUM(i.egreso_cantidad) AS egreso_total
    FROM gastos i
    JOIN estudiante e ON i.id_estudiante = e.id_estudiante
    GROUP BY e.nombre, e.apellido
"""
df_egresos_totales = pd.read_sql(query_egresos_totales, pg_engine)

# Ingresos (ingreso) desde MySQL
df_mysql_gastos = pd.read_sql("SELECT * FROM ingreso", mysql_engine)
print(df_mysql_gastos.head())
```

```python

# ================== FACT: fact_bienestar ==================

fact_bienestar = (
    df_general[["nombre", "apellido", "distancia"]]
    .merge(
        df_bienestar[
            [
                "nombre",
                "apellido",
                "promedio_académico",
                "distancia_universidad",
                "viajes_origen",
                "indice_bienestar",
            ]
        ],
        on=["nombre", "apellido"],
        how="inner",
    )
    .merge(
        df_ingresos_totales[["nombre", "apellido", "ingreso_total"]],
        on=["nombre", "apellido"],
        how="inner",
    )
    .merge(
        df_egresos_totales[["nombre", "apellido", "egreso_total"]],
        on=["nombre", "apellido"],
        how="inner",
    )
)

fact_bienestar["balance_neto"] = (
    fact_bienestar["ingreso_total"] - fact_bienestar["egreso_total"]
)

# ================== fact_bienestar_keys ==================

# Info de universidad / periodo / residencia / origen desde df_general
general_keys = (
    df_general[
        [
            "nombre",
            "apellido",
            "universidad",
            "carrera",
            "semestre",
            "anio",
            "periodo",
            "zona",
            "sector",
            "tipo_vivienda",
            "codigo_postal",
            "ciudad",
            "provincia",
            "longitud",
            "latitud",
        ]
    ]
    .drop_duplicates()
)

# Tiempo de traslado desde df_bienestar
bienestar_keys = (
    df_bienestar[["nombre", "apellido", "tiempo_traslado_min"]]
    .drop_duplicates()
)

fact_bienestar_keys = (
    fact_bienestar
    .merge(general_keys, on=["nombre", "apellido"], how="inner")
    .merge(bienestar_keys, on=["nombre", "apellido"], how="inner")
)

# ================== map_estudiante ==================

map_estudiante = (
    df_general[["identifier", "nombre", "apellido"]]
    .drop_duplicates()
    .rename(columns={"identifier": "id_estudiante"})
)

# ================== fact_bienestar_modelo ==================

# 1) unir con id_estudiante
fact = fact_bienestar_keys.merge(
    map_estudiante, on=["nombre", "apellido"], how="inner"
)

# 2) unir con dimensiones

# dim_universidad: renombrar columna universidad -> nombre_universidad para el join
dim_uni_for_join = dim_universidad.rename(
    columns={"universidad": "nombre_universidad"}
)

fact = fact.merge(
    dim_uni_for_join[["id_universidad", "nombre_universidad", "carrera"]],
    left_on=["universidad", "carrera"],
    right_on=["nombre_universidad", "carrera"],
    how="inner",
)

fact = fact.merge(
    dim_periodo[["id_periodo", "semestre", "anio", "periodo"]],
    on=["semestre", "anio", "periodo"],
    how="inner",
)

fact = fact.merge(
    dim_residencia[
        [
            "id_residencia",
            "zona",
            "sector",
            "tipo_vivienda",
            "codigo_postal",
            "tiempo_traslado_min",
        ]
    ],
    on=["zona", "sector", "tipo_vivienda", "codigo_postal", "tiempo_traslado_min"],
    how="inner",
)

fact = fact.merge(
    dim_origen[
        [
            "id_origen",
            "ciudad",
            "provincia",
            "longitud",
            "latitud",
        ]
    ],
    on=["ciudad", "provincia", "longitud", "latitud"],
    how="inner",
)

# 3) seleccionar columnas finales y renombrar
fact_bienestar_modelo = fact[
    [
        "id_estudiante",
        "id_universidad",
        "id_periodo",
        "id_residencia",
        "id_origen",
        "promedio_académico",
        "indice_bienestar",
        "distancia_universidad",
        "distancia",
        "viajes_origen",
        "ingreso_total",
        "egreso_total",
        "balance_neto",
    ]
].copy()

fact_bienestar_modelo = fact_bienestar_modelo.rename(
    columns={
        "promedio_académico": "promedio_academico",
        "distancia_universidad": "distancia_universidad_km",
        "distancia": "distancia_origen_km",
        "viajes_origen": "num_viajes_origen",
    }
)

# ================== Limpiar dimensiones (solo las que aparecen en el fact) ==================

ids_est = fact_bienestar_modelo["id_estudiante"].unique()
ids_uni = fact_bienestar_modelo["id_universidad"].unique()
ids_per = fact_bienestar_modelo["id_periodo"].unique()
ids_res = fact_bienestar_modelo["id_residencia"].unique()
ids_ori = fact_bienestar_modelo["id_origen"].unique()

dim_estudiante_clean = dim_estudiante[
    dim_estudiante["id_estudiante"].isin(ids_est)
].copy()

dim_universidad_clean = dim_universidad[
    dim_universidad["id_universidad"].isin(ids_uni)
].copy()

dim_periodo_clean = dim_periodo[
    dim_periodo["id_periodo"].isin(ids_per)
].copy()

dim_residencia_clean = dim_residencia[
    dim_residencia["id_residencia"].isin(ids_res)
].copy()

dim_origen_clean = dim_origen[
    dim_origen["id_origen"].isin(ids_ori)
].copy()
```

```python
dim_estudiante = dim_estudiante_clean
dim_universidad = dim_universidad_clean
dim_periodo = dim_periodo_clean
dim_residencia = dim_residencia_clean
dim_origen = dim_origen_clean
```
